```python
def format_follower_count(n):
    if n>= 1000000:
        return f"{n/1000000:.1F}M"
    elif n>= 1000:
        return f"{n/1000:.f}M"
    else:
        return str(n)



print(format_follower_count(1500))
print(format_follower_count(2300000))
print(format_follower_count(850))
                  
        
```
