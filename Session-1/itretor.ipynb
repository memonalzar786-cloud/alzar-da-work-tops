```python
foods = ["Pizza", "Burger", "Pasta", "Sandwich"]

prices = [250, 120, 180, 100]

for food, price in zip(foods, prices):
    print(f"{food}-₹{price}")
```

```python
def insta_posts_generator(posts):
    for post in posts:
        yield post

posts = [
    "Enjoying my day!",
    "Learning Python",
    "Data Analytics Journey",
    "New Post!"
]

post_generator = insta_posts_generator(posts)

try:
    print(next(post_generator))
    print(next(post_generator))
    print(next(post_generator))
    print(next(post_generator))
except StopIteration:
    print("NO more posts")
```

```python
def cashback_genrator(transactions):
    for amount in transactions:
        cashback = amount*0.5
        yield cashback

transactions = [500,1000,2000,3000]

for cashback in cashback_genrator(transactions):
    print("Cashback:", cashback)
```
