```python
add_gst = lambda p : p+(p*0.18)

print(add_gst(100))
print(add_gst(500))
print(add_gst(1000))
```

```python
products = [
    "Shoes",
    "Laptop",
    "Smartphone",
    "Shirt",
    "Watch",
    "Speaker"
]

result = list(filter(lambda p: p.lower().startswith("s"),products))
print(result)
```

```python
from functools import reduce
order = [120, 340, 560, 80]

total_bill = reduce(lambda a,b: a+b, order) 
print(total_bill)
```

```python
from functools import reduce
numbers = [40,60,80,120]

doubled = list(map(lambda x:x*2,numbers))
filterd = list(filter(lambda x:x > 100, doubled))
total = reduce(lambda a,b:a+b,filterd)

print("Doubled:", doubled)
print("Filter:", filterd)
print("Total:", total)
```
