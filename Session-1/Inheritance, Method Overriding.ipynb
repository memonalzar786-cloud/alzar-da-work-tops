```python
class Product:

    def get_discount(self):
        return 0

class Electronics(Product):
        def get_discount(self):
            return 10

Product = Product()
electronics = Electronics()

print("Product Discount:", Product.get_discount())
print("Electronics Discount:",electronics.get_discount())
```

```python
class foodoreder:
    def __init__(self, base_price):
        self.base_price = base_price

    def calculate_total(self):
        return self.base_price

class zoomatorder(foodoreder):
    def calculate_total(self):
        delivery_charge = self.base_price*0.05
        return self.base_price+delivery_charge

order = foodoreder(1000)
zometo = zoomatorder(1000)

print("Normal order:", order.calculate_total())
print("zomato order:", zometo.calculate_total())
    
    
```

```python
class influencer:
    def bonus(self):
        return 2000
class BrandManager:
    def bonus(self):
        return 5000

def show_bonus(employee):
    print("Employee Bonus:", employee.bonus())

influencer= influencer()
manager = BrandManager()

show_bonus(influencer)
show_bonus(manager)
```

```python
class user:
    def get_status(self):
        return "active"
class PremiumUser(user):
    def get_status(self):
        return "Premium"

user = user()
premiuam_user = PremiumUser()

print("User Status:", user.get_status())
print("Premium Useri:", premiuam_user.get_status())
```
