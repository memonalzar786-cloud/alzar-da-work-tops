```python
def safe_divide(a,b):
    try:
        return a/b
    except ZeroDivisionError:
        return "can not divide by zero"

print(safe_divide(10,2))
print(safe_divide(10,0))
```

```python
class invalidDurationerror(Exception):
    pass 

def get_playlist_duration(songs):
    total = 0

    for duration in songs:
        if duration < 0:
            raise invalidDurationerror("song duration can not be negative")

        total += duration
    return total / 60

durations = [180,200,240,120]

try:
    result = get_playlist_duration(durations)
    print("Total Duration:", result, "minutes")

except invalidDurationerror as e:
    print("Error:",e)
```
