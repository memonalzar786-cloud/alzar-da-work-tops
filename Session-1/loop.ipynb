```python
scores = [45, 78, 102, 34, 67, 89]
i = 0

while i < len(scores):
    if scores[i] > 100:
        break

    print(scores[i])
    i = i + 1 
```

```python

prices = [299, 499, 199, 999, 149]

total = 0

for price in prices:
    if price < 200:
        continue

    total = total + price
print("Total:", total)
```

```python
followers_list = [120, 1500, 23000, 800, 45000]

for followers in followers_list:
    if followers < 1000:
        print(followers, "- Micro")
    elif followers <= 10000:
        print(followers, "- Influencer")
    else:
        print(followers, "- Celebrity")
```
