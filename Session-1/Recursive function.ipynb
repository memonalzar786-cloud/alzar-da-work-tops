```python
def reverse_strings(s):
    if s == "":
        return"" 
    return reverse_strings(s[1:])+ s[0]

result = reverse_strings("Heloo")
print(result)   
```

```python
def sum_playlist_duractions(duractions):
    if len(duractions) == 0:
        return 0
    return duractions[0] +sum_playlist_duractions(duractions[1:])

duractions = [180,200,150,240]
total = sum_playlist_duractions(duractions)
print("Total Duraction:", total, "seconds")
```

```python
def count_likes(posts):

    total = posts["likes"]

    for reply in posts.get("replies", []):
        total += count_likes(reply)

    return total


instagram_posts = {
    "likes": 100,
    "replies": [
        {
            "likes": 50,
            "replies": [
                {"likes": 20}
            ]
        },
        {
            "likes": 30
        }
    ]
}

print("Total Likes:", count_likes(instagram_posts))
```
