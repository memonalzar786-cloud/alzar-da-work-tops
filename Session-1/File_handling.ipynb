```python
import csv

with open ("ipl_matches.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["Match", "Team", "Team2", "Winner"])

    writer.writerow([1,"CSK", "MI", "CSK"])
    writer.writerow([2, "RCB", "KKR", "KKR"])
    writer.writerow([3, "GT", "RR", "GT"])

print("csv file created")


```

```python
import csv

with open("ipl_matches.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print("Winning Team:", row["Winner"])
```
