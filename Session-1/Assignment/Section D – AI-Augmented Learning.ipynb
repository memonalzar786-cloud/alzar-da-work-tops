Create a Python Food Delivery Billing Program.

Requirements:
1. Accept multiple food items and their prices from the user.
2. Calculate the itemized subtotal.
3. Add 18% GST and a flat delivery fee of Rs 30.
4. Apply a 10% loyalty discount on the subtotal if the customer has placed more than 5 previous orders.
5. Take previous order count as user input.
6. Print a formatted receipt showing item names, prices, subtotal, GST, delivery fee, discount, and final amount.
7. Handle negative prices and non-numeric order counts using exception handling.

```python



items = []

try:
    count = int(input("How many food items? "))

    for i in range(count):
        name = input("Enter food item name: ")
        price = float(input("Enter price: "))

        if price < 0:
            print("Negative price is not allowed.")
            break

        items.append({"name": name, "price": price})

    else:
        previous_orders = int(input("Enter previous orders: "))

        subtotal = sum(item["price"] for item in items)
        gst = subtotal * 0.18
        delivery_fee = 30

        if previous_orders >= 5:
            discount = subtotal * 0.10
        else:
            discount = 0

        final_amount = subtotal + gst + delivery_fee - discount

        print("\n----- FOOD BILL -----")

        for item in items:
            print(f"{item['name']}: Rs {item['price']:.2f}")

        print(f"Subtotal: Rs {subtotal:.2f}")
        print(f"GST 18%: Rs {gst:.2f}")
        print(f"Delivery Fee: Rs {delivery_fee:.2f}")
        print(f"Discount: Rs {discount:.2f}")
        print(f"Final Amount: Rs {final_amount:.2f}")

except ValueError:
    print("Error: Please enter valid numeric values.")
```
