Que:- 1

The order total Rs 499.50 and delivery distance 7.3 km should be stored as float data types because they contain decimal values. The payment method 'UPI' should be stored as a string because it contains text. The restaurant's availability should be stored as a boolean because it represents either True or False.

If Rs 499.50 is stored as an integer, its decimal part may be lost and the value may become 499. This can produce an incorrect discount calculation. Therefore, float is the appropriate data type for storing prices that contain decimal values.

Que:- 2

A dictionary provides faster and more readable access when looking up a dish price by name. In a dictionary, the dish name is used as a key, so we can directly access its price using the key.

In a list of tuples, we need to search through the list to find the required dish and then access its price using an index. This becomes less efficient as the menu grows.

One limitation of the list-of-tuples approach is that it does not provide direct key-based access. A dictionary solves this problem by allowing us to access menu details directly using the dish name.

Que:- 3

A named function defined with def is more appropriate for the delivery fee calculator because it contains multiple conditions and is easier to read, understand, and maintain. It can use if-elif-else statements to check the order value and delivery distance and return the correct delivery fee.

A lambda function is better for a simple operation in the same app, such as calculating an 18% GST amount on a list of food prices. Lambda is suitable for short, single-expression operations.

Que:- 4

To save a single order dictionary to a JSON file, first import the json module. Then open the file using open() in write mode ('w') and use json.dump() to write the dictionary into the file.

To read the order back, open the JSON file in read mode ('r') and use json.load() to convert the stored JSON data into a Python object.

If the file does not exist, Python raises a FileNotFoundError exception. The program should handle this exception using a try-except block and initialize an empty list of orders. This allows the application to start cleanly on its first run without crashing.

Que:- 5

A class is a better design choice because it groups related data and actions together in one structure. It makes the delivery tracking system easier to organize, maintain, and manage when there are multiple riders.

The first OOP principle is Encapsulation, which combines rider data and methods inside a class. It helps keep related information and operations together.

The second principle is Abstraction, which allows users to perform actions such as updating the location or completing a delivery without needing to understand the internal implementation of those methods.
